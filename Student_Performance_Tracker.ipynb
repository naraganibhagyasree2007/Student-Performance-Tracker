```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
```

```python
data = pd.read_csv("students.csv")
data
```

```python
subjects = ["Python", "Java", "Maths", "AI_ML"]

data["Total"] = data[subjects].sum(axis=1)
data["Percentage"] = (data["Total"] / 400) * 100

data[["Student_ID", "Name", "Total", "Percentage"]]
```

```python
def get_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

data["Grade"] = data["Percentage"].apply(get_grade)

data["Result"] = data["Percentage"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)

data[[
    "Student_ID",
    "Name",
    "Total",
    "Percentage",
    "Grade",
    "Result"
]]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(data["Name"], data["Percentage"])

plt.xlabel("Students")
plt.ylabel("Percentage")
plt.title("Student Performance")
plt.ylim(0, 100)

plt.show()
```

```python
student_id = input("Enter Student ID: ")

student = data[data["Student_ID"] == student_id]

if student.empty:
    print("Student not found.")
else:
    print("\n===== STUDENT PERFORMANCE =====")
    print(student[[
        "Student_ID",
        "Name",
        "Total",
        "Percentage",
        "Grade",
        "Result"
    ]].to_string(index=False))
```

```python
print("===== CLASS PERFORMANCE SUMMARY =====")

print("Number of Students:", len(data))
print("Average Percentage:", round(data["Percentage"].mean(), 2))
print("Highest Percentage:", round(data["Percentage"].max(), 2))
print("Lowest Percentage:", round(data["Percentage"].min(), 2))
print("Students Passed:", (data["Result"] == "Pass").sum())
print("Students Failed:", (data["Result"] == "Fail").sum())
```

```python
print("=" * 60)
print("          STUDENT PERFORMANCE TRACKER")
print("=" * 60)

print("\nSTUDENT PERFORMANCE REPORT\n")

report = data[[
    "Student_ID",
    "Name",
    "Total",
    "Percentage",
    "Grade",
    "Result"
]].copy()

report["Percentage"] = report["Percentage"].round(2)

print(report.to_string(index=False))

print("\n" + "=" * 60)
print("                 CLASS SUMMARY")
print("=" * 60)

print(f"Total Students : {len(data)}")
print(f"Class Average  : {data['Percentage'].mean():.2f}%")
print(f"Highest Score  : {data['Percentage'].max():.2f}%")
print(f"Lowest Score   : {data['Percentage'].min():.2f}%")
print(f"Passed         : {(data['Result'] == 'Pass').sum()}")
print(f"Failed         : {(data['Result'] == 'Fail').sum()}")
```

Student Performance Tracker

Student Performance Tracker is a Python-based data analysis project that helps track and analyze student academic performance.

The system reads student marks from a CSV file and automatically calculates:

• Total marks
• Percentage
• Grade
• Pass/Fail result
• Class performance statistics
• Individual student performance
• Performance visualization

Technologies Used: Python, Pandas, Matplotlib, Google Colab

Project Type: Student Performance Analysis

Objectives

• To store and manage student marks.
• To calculate total marks and percentage automatically.
• To assign grades based on student performance.
• To determine Pass/Fail status.
• To analyze overall class performance.
• To visualize student performance using charts.
• To search and display individual student performance.

Features

• Student data management using CSV.
• Automatic calculation of total marks.
• Automatic percentage calculation.
• Automatic grade assignment.
• Pass/Fail result generation.
• Individual student search by Student ID.
• Class performance summary.
• Student performance visualization using a bar chart.

How the Project Works

1. The system reads student marks from the students.csv file.
2. Pandas is used to load and process the student data.
3. Total marks are calculated for each student.
4. The percentage is calculated from the total marks.
5. A grade is assigned based on the percentage.
6. The student is classified as Pass or Fail.
7. A performance chart is generated using Matplotlib.
8. The system can search for an individual student using Student ID.
9. A class performance summary is generated.

```python
# Create a clean final performance report

final_report = data[[
    "Student_ID",
    "Name",
    "Python",
    "Java",
    "Maths",
    "AI_ML",
    "Total",
    "Percentage",
    "Grade",
    "Result"
]].copy()

final_report["Percentage"] = final_report["Percentage"].round(2)

# Save the processed results
final_report.to_csv("student_performance_report.csv", index=False)

print("Final report created successfully!")
print("\n")
print(final_report.to_string(index=False))
```
